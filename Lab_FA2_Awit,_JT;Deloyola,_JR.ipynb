```r
# ============================================================================
# Data Wrangling and Transformation Assignment
# Dataset: mpg (fuel economy data)
# Based on Lecture 4: Data Wrangling and Transformations

# Step 1: Load the Dataset
# Load tidyverse package which includes ggplot2 (contains mpg dataset)
library(tidyverse)
install.packages("patchwork")
install.packages("moments")
# Load the mpg dataset
data("mpg")

# ----------------------------------------------------------------------------
# Task 1: Data Inspection
# ----------------------------------------------------------------------------

# 1.1 Confirm that mpg is loaded as a tibble
cat("=== Checking if mpg is a tibble ===\n")
is_tibble(mpg)
class(mpg)

# 1.2 Display the first six rows of the dataset
cat("\n=== First six rows of mpg dataset ===\n")
head(mpg)

# Alternative using slice_head()
cat("\n=== Using slice_head() to display first 6 rows ===\n")
mpg %>% slice_head(n = 6)

# Additional exploration: Check structure and dimensions
cat("\n=== Dataset structure ===\n")
glimpse(mpg)

cat("\n=== Dataset dimensions ===\n")
cat("Rows:", nrow(mpg), "\n")
cat("Columns:", ncol(mpg), "\n")

# ----------------------------------------------------------------------------
# Task 2: Data Wrangling and Transformation
# ----------------------------------------------------------------------------

# 2.1 Use separate() to split the 'trans' variable
# The 'trans' variable contains values like "auto(l5)" or "manual(m6)"
# We'll separate it into transmission type and number of gears

cat("\n=== Original trans variable (sample) ===\n")
mpg %>% select(trans) %>% head(10)

# Create a new dataset with separated transmission variables
mpg_wrangled <- mpg %>%
  separate(
    col = trans,                    # Column to separate
    into = c("trans_type", "gears"), # New column names
    sep = "\\(",                     # Separator: opening parenthesis
    remove = FALSE                   # Keep original column
  ) %>%
  mutate(
    gears = str_replace(gears, "\\)", "")  # Remove closing parenthesis
  )

cat("\n=== After separating trans variable ===\n")
mpg_wrangled %>%
  select(trans, trans_type, gears) %>%
  head(10)

# 2.2 Use mutate() to create transformed variables

# Log transformation of hwy (highway fuel efficiency)
# Log transformation is useful for right-skewed data
mpg_transformed <- mpg_wrangled %>%
  mutate(
    log_hwy = log(hwy),              # Natural log of highway mpg
    sqrt_displ = sqrt(displ)          # Square root of engine displacement
  )

cat("\n=== Transformed variables created ===\n")
mpg_transformed %>%
  select(hwy, log_hwy, displ, sqrt_displ) %>%
  head(10)

# Summary statistics for original and transformed variables
cat("\n=== Summary statistics ===\n")
cat("\nOriginal hwy:\n")
summary(mpg_transformed$hwy)

cat("\nLog-transformed hwy:\n")
summary(mpg_transformed$log_hwy)

cat("\nOriginal displ:\n")
summary(mpg_transformed$displ)

cat("\nSquare root displ:\n")
summary(mpg_transformed$sqrt_displ)

# ============================================================================
# PART 2: VISUALIZATION AND INTERPRETATION
# ============================================================================

# ----------------------------------------------------------------------------
# Task 1: Visualization (6 points)
# ----------------------------------------------------------------------------

# Set up plotting parameters for better visualization
theme_set(theme_minimal())

# 2.1 Histogram of original fuel efficiency variable (hwy)
cat("\n=== Creating histogram of original hwy ===\n")

plot1 <- ggplot(mpg_transformed, aes(x = hwy)) +
  geom_histogram(bins = 20, fill = "pink", color = "black", alpha = 0.7) +
  labs(
    title = "Distribution of Highway \nFuel Efficiency (Original)",
    x = "Highway MPG",
    y = "Frequency"
  ) +
  theme_minimal(base_size = 12)

print(plot1)

# 2.2 Histogram of transformed fuel efficiency variable (log_hwy)
cat("\n=== Creating histogram of log-transformed hwy ===\n")

plot2 <- ggplot(mpg_transformed, aes(x = log_hwy)) +
  geom_histogram(bins = 20, fill = "blue", color = "black", alpha = 0.7) +
  labs(
    title = "Distribution of Highway \nFuel Efficiency (Log-Transformed)",
    x = "Log(Highway MPG)",
    y = "Frequency"
  ) +
  theme_minimal(base_size = 12)

print(plot2)

# 2.3 Q-Q plots comparing original and transformed versions
cat("\n=== Creating Q-Q plots ===\n")

# Q-Q plot for original hwy
plot3 <- ggplot(mpg_transformed, aes(sample = hwy)) +
  stat_qq(color = "steelblue", alpha = 0.7) +
  stat_qq_line(color = "red", linewidth = 1) +
  labs(
    title = "Q-Q Plot: Original Highway MPG",
    x = "Theoretical Quantiles",
    y = "Sample Quantiles"
  ) +
  theme_minimal(base_size = 12)

print(plot3)

# Q-Q plot for log-transformed hwy
plot4 <- ggplot(mpg_transformed, aes(sample = log_hwy)) +
  stat_qq(color = "steelblue", alpha = 0.7) +
  stat_qq_line(color = "red", linewidth = 1) +
  labs(
    title = "Q-Q Plot: Log-Transformed Highway MPG",
    x = "Theoretical Quantiles",
    y = "Sample Quantiles"
  ) +
  theme_minimal(base_size = 12)

print(plot4)

# Additional: Side-by-side comparison
library(patchwork)
combined_histograms <- plot1 + plot2 +
  plot_layout(ncol = 2)

combined_qq <- plot3 + plot4 +
  plot_layout(ncol = 2)

print(combined_histograms)
print(combined_qq)

# Statistical tests for normality
cat("\n=== Shapiro-Wilk normality tests ===\n")
cat("\nOriginal hwy:\n")
shapiro.test(mpg_transformed$hwy)

cat("\nLog-transformed hwy:\n")
shapiro.test(mpg_transformed$log_hwy)

# ============================================================================
# PART 2 - TASK 2: INTERPRETATION
# ============================================================================

cat("\n")
cat("=============================================================================\n")
cat("INTERPRETATION OF TRANSFORMATION\n")
cat("=============================================================================\n\n")

interpretation <- "The log transformation compressed the right-skewed data of the original hwy variable,
making the distribution appear more symmetrical and bell-shaped in the second histogram.
In the Q-Q plots, the original data showed significant deviation from the diagonal line
at the upper end, whereas the log-transformed data follows the theoretical normality line
much more closely, indicating improved normality.
"
cat(interpretation)
cat("\n")

# ============================================================================
# PART 3: REFLECTION
# ============================================================================

cat("=============================================================================\n")
cat("REFLECTION\n")
cat("=============================================================================\n\n")

reflection <- "We chose the log transformation for highway MPG because the original data was right-skewed,
and logging is a standard technique to stabilize variance and normalize distributions. The transformation
successfully improved the distribution by reducing the impact of outliers and making the data more linear.
This will be highly beneficial for future regression analysis, as many statistical models assume normality
and constant variance in the error terms to provide accurate p-values and estimates.
"

cat(reflection)
cat("\n")

# ============================================================================
# SUMMARY STATISTICS TABLE
# ============================================================================

cat("=============================================================================\n")
cat("SUMMARY TABLE: ORIGINAL VS TRANSFORMED VARIABLES\n")
cat("=============================================================================\n\n")

summary_table <- mpg_transformed %>%
  summarise(
    Variable = c("hwy (original)", "log_hwy", "displ (original)", "sqrt_displ"),
    Mean = c(mean(hwy), mean(log_hwy), mean(displ), mean(sqrt_displ)),
    Median = c(median(hwy), median(log_hwy), median(displ), median(sqrt_displ)),
    SD = c(sd(hwy), sd(log_hwy), sd(displ), sd(sqrt_displ)),
    Skewness = c(
      moments::skewness(hwy),
      moments::skewness(log_hwy),
      moments::skewness(displ),
      moments::skewness(sqrt_displ)
    )
  )

print(summary_table)

cat("\n=== Assignment Complete ===\n")
cat("All tasks have been completed successfully!\n")
```
